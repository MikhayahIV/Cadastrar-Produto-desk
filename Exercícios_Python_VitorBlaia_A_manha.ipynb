```python
# 21. Busca Binaria
listar_ordenada = [1,3,5,7,9,11,13];
alvo = int(input(f"Lista: {listar_ordenada}\nDigite o numero a buscar: "));
# A busca binária corta a lista pela metade a cada tentativa, o que a
# torna muito mais rápida que procurar item por item.
print("Listas começam pelo indice 0")
inicio,fim = 0, len(listar_ordenada) - 1;
encontrado = False;
while inicio <= fim:
  meio = (inicio+fim)// 2; # // é a divisão que descarta a parte decimal
  if listar_ordenada[meio] == alvo:
    print(f"encontrado na posica {meio}");
    break;
  elif listar_ordenada[meio]<alvo: #elif esta no meio do if e do else, seria um "senao se"
    inicio = meio + 1;
  else:
    fim = meio - 1;
if not encontrado:
  print("Numero nao encotrado na lista");

```

```python
# 22. Criptografia de Cifra de César

# "def" cria uma função: um bloco de código reutilizável que recebe
# entradas (parâmetros) e pode devolver um resultado (return).
def cesar(mesagem,chave):
  resultado = "";

  # A cifra de César desloca cada letra um número fixo de posições no alfabeto.

  for letra in mensagem:
    if letra.isalpha(): #verifica se é uma letra do alfabeto
      codigo = ord(letra); # ord() pega a letra e transforma em numero
      novoCodigo = codigo + chave;
      resultado += chr(novoCodigo); # chr() transforma o numero em caractere
    else:
      resultado +=letra; # espaços e pontuação ficam como estão

  return resultado;

mensagem = input("Digite a mensagem: ");
chave = int(input("Digite a chave: "));


print("Mensagem Criptografada: ", cesar(mensagem,chave));
```

```python
# 23. Descriptografar Cifra de César
def descriptografar(mesagem,chave):
  resultado = "";
  # Descriptografar é o mesmo processo, mas deslocando para o lado oposto.

  for letra in mensagem:
    if letra.isalpha(): #verifica se é uma letra do alfabeto
      codigo = ord(letra); # pega a letra e transforma em numero
      novoCodigo = codigo - chave;
      resultado += chr(novoCodigo); # transforma o numero em caractere
    else:
      resultado +=letra;

  return resultado;

mensagem = input("Digite a mensagem: ");
chave = int(input("Digite a chave: "));


print("Mensagem descriptografada: ", descriptografar(mensagem,chave));
```

```python
# 24. Validação de CPF

def validar_cpf(cpf):
  # Mantém só os dígitos, removendo pontos e traço.
  cpf = ''.join(filter(str.isdigit,cpf));
  # Um CPF válido tem 11 dígitos e não pode ser uma sequência repetida (ex: 111.111.111-11).
  if len(cpf) != 11 or cpf == cpf[0] *11:
    return False;
  soma = sum(int(cpf[i]) * (11 - i) for i in range(10));
  resto = (soma * 10) % 11;
  if resto ==10:
    resto = 0;
  if resto != int(cpf[10]):
    return False;
  return True;

cpf_digitado = input("Digite um cpf(com ou sem pontuacao): ");
print("Cpf valido" if validar_cpf(cpf_digitado) else "cpf invalido");
```

```python
# 30. Jogo da Forca
import random;
palavras_forca = ["python", "programacao", "computador", "algoritmo"];
palavra_secreta = random.choice(palavras_forca);  # sorteia uma palavra da lista
print(palavra_secreta);
letras_descobertas = ["_"] * len(palavra_secreta);  # começa tudo com "_"
tentativas_restantes = 6;
letras_tentadas = [];
while tentativas_restantes > 0 and "_" in letras_descobertas:
    print(" ".join(letras_descobertas));
    print(f"Tentativas restantes: {tentativas_restantes}");
    letra = input("Digite uma letra: ").lower();
    if letra in letras_tentadas:
        print("Você já tentou essa letra!");
        continue  # "continue" pula para a próxima repetição do laço
    letras_tentadas.append(letra);
    if letra in palavra_secreta:
        for i, l in enumerate(palavra_secreta):  # enumerate() dá o índice e o valor juntos
            if l == letra:
                letras_descobertas[i] = letra;
    else:
        tentativas_restantes -= 1;
        print("Letra errada!");
if "_" not in letras_descobertas:
    print(f"Você venceu! A palavra era '{palavra_secreta}'");
else:
    print(f"Você perdeu! A palavra era '{palavra_secreta}'");
```

```python
# 31. Ordenação por Bolha (Bubble SORT)
lista_bubble = [5,3,8,1,9,2];
n = len(lista_bubble);
# A cada volta, os pares vizinhos fora de ordem são trocados de lugar,
# fazendo o maior valor "borbulhar" até o final.
for i in range(n):
  for j in range(0,n-i-1):
    if lista_bubble[j] > lista_bubble[j+1]:
      lista_bubble[j], lista_bubble[j+1] = lista_bubble[j+1], lista_bubble[j];
      print(f"lista ordenada: {lista_bubble}");
```

```python
# 32. Jogo da velha
tabuleiro = [" "] * 9; # 9 posições vazias, representando o tabuleiro 3x3

def mostrar_tabuleiro():
    # Aqui usamos def apenas como organização interna do exemplo
    # também; deixamos assim só para não repetir o mesmo print 1 vez por jogada.
    print(f"{tabuleiro[0]}|{tabuleiro[1]}|{tabuleiro[2]}");
    print(f"{tabuleiro[3]}|{tabuleiro[4]}|{tabuleiro[5]}");
    print(f"{tabuleiro[6]}|{tabuleiro[7]}|{tabuleiro[8]}");

jogador_atual = "X";
jogo_terminou = False;
combinacoes_vitoria = [(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)];
rodadas = 0;
while not jogo_terminou and rodadas < 9:
    mostrar_tabuleiro();
    posicao = int(input(f"Jogador {jogador_atual}, escolha uma posição (0-8): "));
    if tabuleiro[posicao] != " ":
        print("Posição ocupada, tente novamente!");
        continue;
    tabuleiro[posicao] = jogador_atual;
    rodadas += 1;
    for combinacao in combinacoes_vitoria:
        a, b, c = combinacao;
        if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
            mostrar_tabuleiro();
            print(f"Jogador {jogador_atual} venceu!");
            jogo_terminou = True;
            break;
    jogador_atual = "O" if jogador_atual == "X" else "X";
if not jogo_terminou:
    mostrar_tabuleiro();
    print("Empate!");
```

```python
# 37. Funcao recursiva - fatorial
# Uma função é "recursiva" quando ela chama a si mesma para resolver
# um problema menor, até chegar em um caso simples (caso base).
def fatorial_recursivo(n):
  if n <=1:
    return 1;
  return n * fatorial_recursivo(n-1);

numero_fatorial = int(input("Digite um número: "));
print(f"O fatorial de {numero_fatorial} é {fatorial_recursivo(numero_fatorial)}");
```

```python
# 46. Calcular media e devio padrao
import math;
numeros = [4, 8, 15, 16, 23, 42];
media = sum(numeros) / len(numeros);
# Desvio padrão mede o quanto os valores costumam se afastar da média.
# Fórmula: raiz quadrada da média dos quadrados das diferenças para a média.
soma_quadrados = sum((n - media) ** 2 for n in numeros);
desvio_padrao = math.sqrt(soma_quadrados / len(numeros)); #math.sqrt serve para raiz quadrada
print(f"Média: {media:.2f}");
print(f"Desvio padrão: {desvio_padrao:.2f}");
```

```python
# 48. Ordenacao por insercao (insertion sort)
lista = [5,3,8,1,9,2];
# A cada passo, pegamos um elemento e o "inserimos" na posição correta
# entre os elementos já ordenados à esquerda dele.
for i in range(1,len(lista)):
  chave = lista[i];
  j = i - 1;
  while j >= 0 and lista[j] > chave:
    lista[j+1] =  lista[j];
    j -=1;
  lista[j+1] = chave;
print(lista);
```
